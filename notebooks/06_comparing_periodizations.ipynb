# 6. Two, three or four stages? Comparing chronologies against each other

Scholars disagree about how many stages the Qur'an's composition went through:

- **Two:** Meccan and Medinan, as in Islamic tradition (the Cairo edition) and Nöldeke's outer frame.
- **Three:** Dost's R / AR / A phases (notebook 5). Farrin (2019) also proposes three, Early Meccan / Later Meccan / Medinan, from verse lengths. We don't have Farrin's surah list in machine-readable form yet, so we stand in a scheme of the same *shape*: Nöldeke with Meccan II and III merged.
- **Four:** Nöldeke's Meccan I, II, III and Medinan.

This notebook puts them side by side on the same data. It also adds two kinds of comparison scheme:

- **Data-driven cuts:** sort the surahs by mean verse length and find the best place(s) to cut that list into 2, 3, 4 or 5 stages. These show what verse length alone would suggest.
- **"Early break" variants:** Nöldeke's Meccan I vs everything later, and Dost's R-Qur'an vs everything later.

### The fairness problem

More stages always fit better, in the way a map with more regions always describes the territory more closely. Every score here therefore corrects for the number of stages:

- **ΔBIC** (Bayesian information criterion): fit minus a fixed penalty for every extra parameter. Lower is better, 0 is the best scheme, and a difference of 10 or more counts as strong evidence.
- **z vs random:** how many standard deviations better the scheme fits than random groupings of the same sizes.
- **Weighted kappa:** chance-corrected accuracy at predicting each scheme's stages for held-out surahs (0 = chance, 1 = perfect), regardless of how many stages there are.

### The circularity problem

Every scheme was built partly from verse length or divine names. So we judge them on something none of them used: **the frequencies of particles** (prepositions, conjunctions, negations, the article, attached pronouns), with divine names removed.

## 6.1 How similar are the schemes to each other?

The **adjusted Rand index** compares two groupings: 1 means identical, 0 means no more similar than chance.

```python
pz.agreement_matrix().round(2)
```

Cairo and Nöldeke's two-stage division are nearly the same grouping (0.85). Dost's three phases are close to Nöldeke merged to three (0.86). The verse-length cuts agree only moderately with any scholarly scheme.

## 6.2 Which scheme does the style of the text support best?

```python
X = pz.feature_space("function_words")
comp = pz.compare_schemes(X)
comp[["scheme", "k", "description", "delta_bic", "r2", "z_vs_random"]].round(2)
```

Read this table carefully; it holds several findings at once.

1. **Every scheme beats random**, but the two Meccan/Medinan schemes (Cairo, Nöldeke 2) only just. They explain about 2% of the variation in particle use, while the 3- and 4-stage schemes explain 7–10%.
2. **The best 2-stage split is early, not at the hijra.** The verse-length cut, Nöldeke's Meccan I vs rest and Dost's R-Qur'an vs rest all beat the Meccan/Medinan split by a wide margin. In how the text uses small grammatical words, the sharpest change comes *within* the Meccan period, between the earliest surahs and the rest.
3. **Among 3- and 4-stage schemes**, Nöldeke merged to three, Dost and the 3-way verse-length cut are about equally good. Nöldeke's fourth stage (splitting Meccan II from III) improves the fit, but not by enough to pay BIC's price for the extra parameters.

Check how robust this is by dropping the shortest surahs, whose particle frequencies are the noisiest:

```python
comp100 = pz.compare_schemes(pz.feature_space("function_words", min_words=100))
comp100[["scheme", "k", "delta_bic", "r2", "z_vs_random", "n_surahs"]].round(2)
```
